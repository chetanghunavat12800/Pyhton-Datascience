### Group by the aggregation

````python
import numpy as np
import pandas as pd 
````

````python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
    'Store': ['S1', 'S1', 'S2', 'S2', 'S1', 'S2', 'S2', 'S1'],
    'Sales': [100, 200, 150, 250, 120, 180, 200, 300],
    'Quantity': [10, 15, 12, 18, 8, 20, 15, 25],
    'Date': pd.date_range('2023-01-01', periods=8)
}

df1 = pd.DataFrame(data)
print(df1)
````

````python
# group by category and calculate the sum of sales per catgegory
cat=df1.groupby('Category')
for i,v in cat:
    print(i)
    print(v)

cat
````

````python
df1.groupby('Category')['Sales'].sum()
````

````python
total_sales=df1['Sales'].sum()
print(total_sales)
````

````python
df1.groupby('Store')['Sales'].sum()
````

````python
# group by multiple columns
xyz=df1.groupby(['Store','Category'])
for i,v in xyz:
    print(i)
    print(v)
    
````

````python
mean_sales=df1['Sales'].mean()
print(mean_sales)
````
