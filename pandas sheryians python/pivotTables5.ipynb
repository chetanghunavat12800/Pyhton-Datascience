```python
import pandas as pd
import numpy as np
```

### Pivot Tables and aggregation functions AND cross tab

```python
data = {
    'Date': pd.date_range('2023-01-01', periods=20),
    'Product': ['A', 'B', 'C', 'D'] * 5,
    'Region': ['East', 'West', 'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West',
               'North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South'],
    'Sales': np.random.randint(100, 1000, 20),
    'Units': np.random.randint(10, 100, 20),
    'Rep': ['John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary',
           'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice', 'John', 'Mary', 'Bob', 'Alice']
}

pd.DataFrame(data)

```

```python
df=pd.DataFrame(data)
print(df)

df['Month']=df['Date'].dt.month_name()
df['Quarter']='Q'+df['Date'].dt.quarter.astype(str)
df
```

```python
pd.pivot_table(df,values='Sales',index='Region',columns='Product')
```

```python
pd.pivot_table(df,values='Sales',index='Region',columns='Product',aggfunc='sum')    # here agg function is sum it gives the total sales in the particular region for the particular product 
```

```python
# multivalues pivot table 
pd.pivot_table(df,values=['Sales','Units'],index='Region',columns='Product',aggfunc='mean')
```

```python
# cross tab ----it is used for the counting purpose 
pd.crosstab(df['Region'],df['Product'])
```
