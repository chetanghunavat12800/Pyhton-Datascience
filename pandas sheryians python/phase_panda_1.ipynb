```python
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd 
```

## introduction to series
#### series is the 1d data
#### series is in the vertical format 

```python
labels=['a','b','c','d']
my_list=[10,20,30,40]
arr=np.array([100,200,300,400])
dic={1:1000,2:2000,3:3000,4:4000}

# now print the series 
print(pd.Series(my_list))
print("\n")
print(pd.Series(arr))
print("\n")


# if we want to add the custom lables 
print(pd.Series(arr,index=labels))

print()
print(pd.Series(dic))

```

```python
tup=('chetan','cse2','prerna','it')
print(pd.Series(tup))
```

# DATAFRAME 
### DATAFRAME IS MADE UP OF MANY SERIES

```python
data={
    'Names':['chetan','lokendra','abhay','nitin','ronak'],
    'Age': [20,21,21,20,17],
    'City':['sawaimadhopur','orai','bhilai','bhatinda','jodhpur'],
    'Salary':[10,5,8,9,7]
}

label=['A','B','C','D','E']

df=pd.DataFrame(data,index=label)
print(df)
```

```python
data2=[
    ['chetan',20,'sawaimadhopur',10],
    ['prerna',19,'nepanagar',11],
    ['lokendra',21,'orai',5],
    ['abhay',21,'bhilai',8],
    ['nitin',20,'bhatinda',9],
    ['ronak',17,'jodhpur',7]
]
label2=['A','B','C','D','E','F']

df2=pd.DataFrame(data2,index=label2,columns=['Names','Age','city','salary'])
print(df2)


print()

data3=(
    ('chetan',20,'sawaimadhopur',10),
    ('prerna',19,'nepanagar',11),
    ('lokendra',21,'orai',5),
    ('abhay',21,'bhilai',8),
    ('nitin',20,'bhatinda',9),
    ('ronak',17,'jodhpur',7)
)

df3=pd.DataFrame(data3,index=label2,columns=['Names','Age','city','salary'])
print(df3)
```
