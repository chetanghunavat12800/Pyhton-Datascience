```python
import numpy as np
import matplotlib.pyplot as plt 
import pandas as pd 
```

## introduction to series
#### series is the 1d data
#### series is in the vertical format 

```python
labels=['a','b','c','d']
my_list=[10,20,30,40]
arr=np.array([100,200,300,400])
dic={1:1000,2:2000,3:3000,4:4000}

# now print the series 
print(pd.Series(my_list))
print("\n")
print(pd.Series(arr))
print("\n")


# if we want to add the custom lables 
print(pd.Series(arr,index=labels))

print()
print(pd.Series(dic))

```

```python
tup=('chetan','cse2','prerna','it')
print(pd.Series(tup))
```

# DATAFRAME 
### DATAFRAME IS MADE UP OF MANY SERIES

```python
data={
    'Names':['chetan','lokendra','abhay','nitin','ronak'],
    'Age': [20,21,21,20,17],
    'City':['sawaimadhopur','orai','bhilai','bhatinda','jodhpur'],
    'Salary':[10,5,8,9,7]
}

label=['A','B','C','D','E']

df=pd.DataFrame(data,index=label)
print(df)
```

```python
data2=[
    ['chetan',20,'sawaimadhopur',100],
    ['prerna',19,'nepanagar',110],
    ['lokendra',21,'orai',50],
    ['abhay',21,'bhilai',80],
    ['nitin',20,'bhatinda',90],
    ['ronak',17,'jodhpur',70]
]
label2=['A','B','C','D','E','F']

df2=pd.DataFrame(data2,index=label2,columns=['Names','Age','City','Salary'])
print(df2)


print()

data3=(
    ('chetan',20,'sawaimadhopur',100),
    ('prerna',19,'nepanagar',110),
    ('lokendra',21,'orai',50),
    ('abhay',21,'bhilai',80),
    ('nitin',20,'bhatinda',90),
    ('ronak',17,'jodhpur',70)
)

df3=pd.DataFrame(data3,index=label2,columns=['Names','Age','City','Salary'])
print(df3)
```

### slicing and indexing and selecting the data

```python
print(type(df2))
print(df2['Names'])
print()

# if i want to select multiple columns

print(df2[['Names','City']])
```

```python
# inserting the columns in the data 

df2["Batch-Exp"]=2029
print(df2)

print()
df2.insert(2,"Gender",["Male","Female","Male","Male","Male","Male",])  # this insert the gender at the index 2 
print(df2)
```

```python
# if we want to delete row or column 

print(df2.drop('F',axis=0)) # delete the row having index 'F'

print()

print(df2.drop('Batch-Exp',axis=1)) # delete the col having index 'Batch-Exp'

```

```python
df2 # still the data remain same ...
```

```python
# we have to use the inplace=true 

df2.drop('F',axis=0,inplace=True)
df2.drop('Batch-Exp',axis=1,inplace=True)

print(df2)
```

### now selecting the data of the rows

```python
# here we use loc for the rows to give the location of the rows
print(df2.loc['A'])
df2.loc[['A','B']]
```
