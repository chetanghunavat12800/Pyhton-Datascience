```python
import pandas as pd
import numpy as np
```

### finding the missing data in the table 

```python
data=[
    ['chetan',20,np.nan,100],
    ['prerna',19,'nepanagar',110],
    ['lokendra',np.nan,'orai',50],
    [np.nan,21,'bhilai',80],
    ['nitin',20,'bhatinda',np.nan],
    ['ronak',np.nan,'jodhpur',np.nan]
]

label=['A','B','C','D','E','F']
df=pd.DataFrame(data,index=label,columns=['Names','Age','City','Salary'])
print(df)
print()
print(df.isna())
print()
print(df.isna().sum())
print()
print(df.isna().any())
```

```python
# Remove the missing data

df.dropna() # this will drop the all rows having the null value nan
```

```python
df.dropna(thresh=3)
# matlb jis row mai atleast 3 se jyada non null values hai vo delete nahi hoga 
```

```python
# filling the missing data 
df.fillna(0)
```
