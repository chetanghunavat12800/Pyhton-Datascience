### MERGING THE TWO DATAFRAMES ....

```python
import numpy as np 
import pandas as pd
```

```python
emp={
    'employee_id':[1,2,3,4,5],
    'name':['Chetan','Prerna','Lokendra','Nitin','Pranav'],
    'department':['Hr','IT','Finance','Sales','Marketing']
}

sal={
    'employee_id':[1,2,3,5,6,7],
    'salary':[75000,80000,55000,58000,90000,100000],
    'bonus':[7000,4000,5000,6000,8000,9000]
}

employee=pd.DataFrame(emp)
salary=pd.DataFrame(sal)
print(employee,"\n\n",salary)
```

```python
pd.merge(employee,salary) # yha pe woh hi merge ho rahi hai jo dono mai coommon hai
```

```python
# inner and outer merge
pd.merge(employee,salary,on='employee_id',how='inner')
```

```python
merded_df=pd.merge(employee,salary,on='employee_id',how='outer')
print(merded_df)
# here you will get all the employers and with the data and if any data is missing then there is null value NaN
```

```python
# now replace null value by '___' 
values={
    'employee_id':00,
    'name':'unknown',
    'department':'NOT_found',
    'salary':'nil',
    'bonus':'Nil'
    
}
merded_df.fillna(value=values)
```

```python
pd.merge(employee,salary,on='employee_id',how='left')
```

```python
pd.merge(employee,salary,on='employee_id',how='right')
```

## now how to concatenate the two dataframes 

```python
import pandas as pd

df1 = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Priya", "Rahul", "Aman", "Chetan"],
    "Age": [20, 21, 22, 20],
    "City": ["Jaipur", "Delhi", "Indore", "SawaiMadhopur"],
    "Salary": [38000, 35000, 42000,50000 ]
})

df2 = pd.DataFrame({
    "ID": [105, 106, 107, 108],
    "Name": ["Prerna", "Rohit", "Anjali", "Vikas"],
    "Age": [20, 23, 21, 24],
    "City": ["Nepanagar", "Mumbai", "Pune", "Kota"],
    "Salary": [55000, 45000, 36000, 52000]
})



df3=pd.concat([df1,df2],ignore_index=True)
print(df3)
```

```python
df4=pd.DataFrame({
    'Work-Location':['Remote','On-site','On-site','hybrid','Remote','hybrid','remote','remote'],
    'Department':['Cse','IT','Ec','Cse','IT','Cse','AI','Ec']
}
)


print(df4)
```

```python
# concatenate df3 and df4
p=pd.concat([df3,df4],axis=1)
print(p)
```

```python
# joining 2 data frames 
dfr1=pd.DataFrame({
    'name':['Chetan','Prerna','Lokendra','Nitin'],
},index=[1,2,3,4])

dfr2=pd.DataFrame({
    'ID':[101,102,104,105]
},index=[1,2,4,5])

dfr1.join(dfr2)
```

```python
dfr2.join(dfr1)
```

```python
dfr1.join(dfr2,how='outer')
```

```python
dfr2.join(dfr1,how='outer')
```
