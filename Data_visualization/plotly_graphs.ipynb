```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns 
import plotly.express as px
import plotly.io as pio
pio.renderers.default = "browser"
```

```python
# plotly express.....
df = sns.load_dataset("tips")

fig = px.scatter(
    data_frame=df,
    x="total_bill",
    y="tip",
    title="Restaurant Bill vs Tip"
)

fig.show()
```

```python
# make it more interactive 
df = sns.load_dataset("tips")

fig = px.scatter(
    data_frame=df,
    x="total_bill",
    y="tip",
    color='sex',
    size='size',
    title="Restaurant Bill vs Tip"
)

fig.show()

# # make it more interactive 
# df = sns.load_dataset("tips")

# px.scatter(
#     data_frame=df,
#     x="total_bill",
#     y="tip",
#     color='sex',
#     size='size',
#     title="Restaurant Bill vs Tip"
# )
```
