```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df=pd.read_csv('IPL.csv')
df
```

```python
df.head()
```

```python
df.info()
```

```python
df.isna().sum()  # here all of the columns sum if is 0 then there is no null values in the dataframe 
```

```python
# which team won the most matches
df['counts']=1
df.groupby('match_winner')['counts'].sum().sort_values(ascending=False) ## by chetan 
```

```python
# matches winn by each city
match_win=df['match_winner'].value_counts()
match_win=pd.DataFrame(match_win)
match_win
```

```python
# if we want to see won by wickets and runs
match_win['win_by_wickets']=df.loc[df['won_by']=='Wickets']['match_winner'].value_counts()
match_win['win_by_runs']=df.loc[df['won_by']=='Runs']['match_winner'].value_counts()
match_win
```

```python
fig=plt.figure(figsize=(10,7))
sns.countplot(x=df['match_winner'],width=0.7)
```

```python
# barplot 
sns.barplot(data=match_win,x='count',y=match_win.index,palette='viridis')
```

### pie chart of the data

```python
colours=sns.color_palette('tab20',20)
# colours=sns.color_palette('husl',40)
plt.subplot(1,2,1)
plt.pie(x=match_win['count'],labels=match_win.index,colors=colours,autopct='%1.2f%%',radius=1.5)
plt.show()
# to see the actual data else percentage.....

x=match_win['count'].sum()
def my_pct(val):
    a=(val/100)*x
    return a.round()
plt.subplot(1,2,2)
plt.pie(x=match_win['count'],labels=match_win.index,colors=colours,autopct=my_pct,radius=1.5)
plt.show()
```
