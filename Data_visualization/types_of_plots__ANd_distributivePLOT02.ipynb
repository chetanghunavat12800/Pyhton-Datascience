```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df=sns.load_dataset('tips')
df
```

### making histplot of the quantitative entities

```python
sns.histplot(df['total_bill'],bins=10,kde=True) # kernel density estimater 
```

### joinplot to compare the two graphical datas 

```python
sns.jointplot(data=df,x='total_bill',y='tip',kind='scatter')
# this is the scatter point graph 
```

```python
sns.jointplot(data=df,x='total_bill',y='tip',kind='hist')
sns.jointplot(data=df,x='total_bill',y='tip',kind='kde')
sns.jointplot(data=df,x='total_bill',y='tip',kind='hex')
sns.jointplot(data=df,x='total_bill',y='tip',kind='reg')
sns.jointplot(data=df,x='total_bill',y='tip',kind='resid')
```

### pairplot 

```python
sns.pairplot(df)
```

```python
sns.pairplot(df,hue='sex')
```
