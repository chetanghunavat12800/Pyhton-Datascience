### basics of the matplotlib

```python
import numpy as np 
import pandas as pandas
import matplotlib.pyplot as plt 
```

```python
t=np.linspace(-6,6,11)
t
```

```python
s=(t**2)
s
```

```python
plt.plot(t,s)
plt.title('t v/s s graph')
plt.xlabel('t')
plt.ylabel('s')
```

```python
x=np.arange(-10,11,1)
x
```

```python
# y=[]
# for i in x:
#     if i<0:
#         y.append(-i)
#     else:
#        y.append(i)
        
y=np.abs(x)
y
```

```python


plt.subplot(2,2,1)
plt.plot(x,y)
plt.axhline(0)
plt.axvline(0)

plt.subplot(2,2,2)
plt.plot(t,s)
plt.axhline(0)
plt.axvline(0)

plt.subplot(2,2,3)
plt.plot(y,x)
plt.axhline(0)
plt.axvline(0)

plt.subplot(2,2,4)
plt.plot(s,t)
plt.axhline(0)
plt.axvline(0)

```

```python
fig=plt.figure()
axis1=fig.add_axes([0.5,0.1,0.5,0.7])
axis1.plot(t,s)
axis1.plot(x,y)
plt.axhline(0, color="green", linestyle="--",linewidth=0.7)
plt.axvline(0, color="green", linestyle="--",linewidth=0.7)
```
